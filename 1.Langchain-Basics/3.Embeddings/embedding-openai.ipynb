```python
embeddings_1024=OpenAIEmbeddings(model="text-embedding-3-large",dimensions=1024)
embeddings_1024
```

## Understanding Embedding Dimensions

When we run `len(query_result)` and get `1024`, this represents the dimensions of the **entire text chunk as a single vector**, not each individual word. 

The OpenAI embedding model processes the full sequence of words together to capture the overall context. It then compresses that semantic meaning—whether the input is a single word, a sentence, or a whole paragraph—into one single 1024-dimensional array. This single vector is what gets stored and queried inside your Vector Database (like ChromaDB).

```python
text="this is a tutorial on OPENAI embeddings"

query_results=embeddings_1024.embed_query(text)
len(query_results)
```

```python
from langchain_community.document_loaders import TextLoader

loader=TextLoader('speech.txt')
docs=loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
final_documents=text_splitter.split_documents(docs)
```

## Vector Embedding and Vector StoreDB

```python
from langchain_community.vectorstores import Chroma

db=Chroma.from_documents(final_documents,embeddings_1024)
db
```

```python
## Retrieve the results from vector db
query="It will be all the easier for us to conduct ourselves as belligerents"
retrieved_results=db.similarity_search(query)
print(retrieved_results)
```
